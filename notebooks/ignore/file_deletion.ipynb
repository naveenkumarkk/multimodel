```python
import os
import random
```

```python
folder_path = r"..\data\specific_cancer_dataset\train\brain"
```

```python
letters = ['G','M','N','O','P']
```

```python
numbers = list(range(1, 801))
```

```python
num_series_to_keep = 150
```

```python
selected_series = []
```

```python
while len(selected_series) < num_series_to_keep:
    letter = random.choice(letters)
    number = random.choice(numbers)
    prefix = f"{letter}_{number}"

    if any(f.startswith(prefix) for f in os.listdir(folder_path)):
        if prefix not in selected_series:
            selected_series.append(prefix)
```

```python
print(f"Selected series to keep: {selected_series}")
```

```python
all_files = [f for f in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, f))]
```

```python
import pandas as pd
import numpy as np
```

```python
all_files
```

```python
for f in all_files:
    if not any(f.startswith(prefix) for prefix in selected_series):
        os.remove(os.path.join(folder_path, f))
        print(f"Deleted: {f}")
```

```python
import os
import random
```

```python
folder_path = r"..\data\specific_cancer_dataset\val\unspecified"
```

```python
num_to_keep = 800
```

```python
all_files = [f for f in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, f))]
```

```python
files_to_keep = set(random.sample(all_files, min(num_to_keep, len(all_files))))
```

```python
for f in all_files:
    if f not in files_to_keep:
        os.remove(os.path.join(folder_path, f))
        print(f"Deleted: {f}")
```
