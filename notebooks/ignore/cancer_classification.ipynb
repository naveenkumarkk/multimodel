```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/cancer_classification.csv')
```

```python
df.head()
```

```python
df.duplicated().sum()
```

```python
df.describe(include='number')
```

```python
df.info()
```

```python
numerical_columns = [var for var in df.columns if df[var].dtype!='O']
```

```python
numerical_columns
```

```python
fig, axes = plt.subplots(1,3,figsize=(20,6))
bg_color = '#FFF5E1'
column = numerical_columns[0]
plot_color = '#FFD1A4'
font_color = '#FF4500'

sns.kdeplot(df[column],ax=axes[0],fill=True,color=plot_color,alpha=0.8,linewidth=0)
axes[0].set_title('Radius Mean plot',fontsize= 13,color=font_color)
axes[0].set_facecolor(bg_color)
axes[0].grid(True,linestyle='--',alpha=0.3)

axes[1].hist(df[column],bins=30,color=plot_color,edgecolor='white',alpha=0.9)
axes[1].set_title('Histogram',fontsize=13,color=font_color)
axes[1].set_facecolor(bg_color)
axes[1].grid(True,linestyle='--',alpha=0.3)

sns.boxplot(x=df[column],ax=axes[2],color=plot_color,fliersize=5,linewidth=2,boxprops=dict(alpha=0.8))
axes[2].set_title('Box Plot',fontsize=13,color=font_color)
axes[2].grid(True,linestyle='--',alpha=0.3)

plt.tight_layout()
plt.show()
```
